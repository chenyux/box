## SOLUTION

```python
# Calculate the sample size
n = employees['age'].count()
```

```python
# Calculate the mean of age
mean_age = employees['age'].mean()
```

```python
# Calculate the standard deviation of age
std_age = employees['age'].std()
```

```python
# Calculate the standard error of the mean (SEM)
SEM = std_age / np.sqrt(n)
```

```python
# Set the confidence level to 0.95
conf = 0.95
```

```python
# Calculate the 95% confidence interval for the mean age using scipy
confidence_interval = stats.norm.interval(confidence=conf, loc=mean_age, scale=SEM)
```
