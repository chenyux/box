## Exercise 3: If/elif/else statements

In this exercise, you will be more specific and check not only whether you are making money or not, but also if you are breaking even.

### Instructions

You have a variable **profit** already defined.

- If **profit** is larger than zero, print "We are making money!".
- If **profit** equals zero, print "We are breaking even!".
- If none of these conditions are met, print "We are losing money:("

````python
profit = 0
````

**Expected output:**

We are breaking even!

## SOLUTIONS

### Exercise 1:

```python
profit = 223
if profit > 0:
    print("We are making money!")
else:
    print("We are not making any money:(")
```

### Exercise 2

```python
profits = [100, -150, 200, 430]
for profit in profits:
    if profit > 0:
        print("We are making money!")
    else:
        print("We are not making any money:(")
```

### Exercise 3:

```python
profit = 0

    print("We are making money!")
```
