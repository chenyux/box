# Dictionaries and NumPy arrays

In this reading item, you will explore two foundational data structures in Python: dictionaries and NumPy arrays. While you'll primarily work with Pandas DataFrames and Series in this course, understanding these structures will enhance your grasp of data organization and manipulation in Python.

## Python Dictionaries

Dictionaries in Python are collections of key-value pairs, similar to real-world dictionaries where you look up a word to find its definition. In a Python dictionary, the key is used to access the corresponding value. This structure is ideal for modeling meaningful relationships in your data.

Creating a Dictionary:

```python
# Dictionary of library visits by year
library_visits = {
    1996: 15000,
    2000: 18000,
    2005: 22000
}
# Accessing a value using a key
print(library_visits[1996])  # Output: 15000
```

Example with Strings as Keys:

```python
# Dictionary of library income by library name
library_income = {
    'Avon': 685678,
    'Central': 1234567,
    'Westside': 987654
}
# Accessing a value using a key
print(library_income['Avon'])  # Output: 685678
```

Adding a New Key-Value Pair:

```python
# Dictionary of library visits by year
library_visits = {
    1996: 15000,
    2000: 18000,
    2005: 22000
}
# Adding a new key-value pair to the dictionary
library_visits[2010] = 25000
print(library_visits)  # Output: {1996: 15000, 2000: 18000, 2005: 22000, 2010: 25000}
```

Accessing Dictionary Keys:

```python
library_income = {
    'Avon': 685678,
    'Central': 1234567,
    'Westside': 987654
}

# Accessing all keys in the dictionary
keys = library_income.keys()
print(keys) # Output: dict_keys(['Avon', 'Central', 'Westside'])

# Useful for iterating over keys or checking for the existence of a key
for key in library_income.keys():
    print(f"{key} made {library_income[key]} in 1996")
```

The **dict.keys()** method returns a view object that displays a list of all the keys in the dictionary. This is useful for iterating over keys or checking if a particular key exists in the dictionary.

Dictionaries are excellent for quick lookups using keys, such as finding library income by library name. However, they are less convenient for performing operations on all keys or values simultaneously. Note that the keys need to be unique: you cannot have two entries with the same key, just like a list cannot have two elements in one place.

## NumPy Arrays

NumPy is a powerful Python library that provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on these arrays. NumPy arrays are similar to lists but offer enhanced performance and capabilities, particularly through vectorized operations.

Creating a NumPy Array:

```python
import numpy as np

# Creating a NumPy array
array = np.array([1, 2, 3, 4, 5])

# Accessing elements
print(array[0]) # Output: 1

# Vectorized operation: adding 10 to each element
print(array + 10) # Output: [11 12 13 14 15]
```

NumPy arrays are efficient for numerical computations and are the foundation of Pandas Series and DataFrames. They allow for vectorized operations, meaning you can apply an operation to all elements in the array simultaneously, rather than iterating through them one by one. Another important difference from lists is that all entries in a NumPy array must be of the same type, so you cannot mix strings with integers and floats just like in lists.

If you have data stored in a list, you can simply transform it to an array for further processing:

```python
import numpy as np

# Transforming a list to a NumPy array
x_list = [10, 20, 30, 40]
x_array = np.array(x_list)
print(x_array) # Output: [10 20 30 40]
```

## Comparing Data Structures

In the image below you can see how the different types build on each other (look at the arrows and compare the properties of each data type). Below the image you can see a textual description.

![dictionaries-and-numpy-arrays](dictionaries-and-numpy-arrays-1.png)

**Python List**: An ordered, one-dimensional data structure where elements of different types can be stored together, accessed using numerical indices starting from zero.

**NumPy Array**: A more powerful version of a list, enabling vectorized operations for efficient mathematical computations, typically allowing only one type of data.

**Dictionary**: A data structure allowing access to values using keys rather than numerical indices.

**Pandas Series**: Combines features of arrays and dictionaries. An ordered, one-dimensional data structure that can have custom indices for key-based access, built on NumPy arrays for vectorization and statistical methods.

**Pandas DataFrame**: A two-dimensional data structure with rows and columns, where each column is a Series, providing all the power and flexibility of Series in two dimensions.
