## Exercise 1: Printing elements of a list

### Instructions

Given a list of names (**names**)

- Write a for loop that iterates over all the elements in **names**
- In each iteration print the text "Hello, " followed by the corresponding name.

````python
names = ["Susan", "Hiroshi", "Zainab"]
````

**Hint 1:** To set up the for loop you need the structure **for** <name_variable> **in names:**. Don't forget the `:` !

**Hint 2:** Remember that all the lines of code that need to happen inside each iteration of the for loop need to be indented!

**Hint 3:** For the print you can use the expression **print("Hello, ",** <current_name> **)**

## Exercise 3: Iterating over two lists

### Instructions

You have two lists: **names** and **ages**.

- Write a for loop that iterates over both lists at once. Use **range()** to achieve that.
- For each iteration in the loop, check if the person is older than 40.
- If the person is older than 40, print out their name.

````python
names = ["Susan", "Hiroshi", "Zainab", "Sade"]
ages = [19, 41, 36, 62]
````

**Hint 1:** use **len()** of one of the lists as an argument to the **range()** function.
