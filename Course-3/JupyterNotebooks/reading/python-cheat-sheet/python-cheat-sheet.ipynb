## Functions

### Pre-defined Functions:

- **len():** Get length.
- **sum():** Sum of elements.
- **min():** Minimum value.
- **max():** Maximum value.

```python
scores = [5, 3, 1, 2, 4]
print(len(scores))  # 5
print(sum(scores))  # 15
print(min(scores))  # 1
print(max(scores))  # 5
```

## Control Flow

Conditional statements in Python, using the **if** keyword, allow you to execute code based on certain conditions. They enable decision-making in your programs by running different blocks of code depending on whether a condition is true or false.

```python
temperature = 18
if temperature > 0:
    print("The water does not freeze.")
else:
    print("The water freezes.")
```

Loops in Python, such as the **for** loop, enable you to execute a block of code multiple times. They are essential for tasks that require repetition, like iterating over a list or performing an action until a condition is met.

- looping over a list of items:

```python
cities = ["Toronto", "Bangalore", "Nairobi", "Paris"]
for city in cities:
    print(city, " is a cool city!")
```

The **range()** function in Python generates a sequence of numbers, which is commonly used in for loops.

- looping over a range:

```python
for i in range(5):
    print(i)
```

**Order of Execution:** Code runs from top to bottom. Loops and conditionals alter this flow.
