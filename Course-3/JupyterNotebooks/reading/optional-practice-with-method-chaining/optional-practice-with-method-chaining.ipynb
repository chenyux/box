## Exercise 1: Rewrite the code by chaining the methods instead of using them one by one

You are part of an HR team preparing an employee spotlight feature for your company newsletter. You're tasked with highlighting employees older than 30, sorted by their years of service, and adding a new column to display how many years they've been in the company so far. To streamline the process and make the data reader-friendly, you decide to rename the "tenure" column to "years_in_company" before printing the final data.

### Instructions

- The code is already provided for you.
- Rewrite the code to use method chaining rather than defining a new variable after using each method. For example you should rewrite this code:
  ```python
  result1 = df.method1()
  result2 = result1.method2()
  final_result = result2.method3()
  ```
  to look like this:
  ```python
  final_result = df.method1().method2().method3()
  ```
- Print the result to check for consistency.

````python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Alice Smith", "Bob Johnson", "Charlie Williams",
             "Diana Brown", "Ethan Jones"],
    "age": [28, 45, 34, 29, 50],
    "job_title": ["Data Analyst", "Senior Developer", "Project Manager", "HR Specialist", "Director"],
    "tenure": [3, 10, 1, 5, 0]
})

### START CODE HERE ###

# Filtering
filtered_employees = employees[employees["age"] > 30]

# Sorting
sorted_employees = filtered_employees.sort_values("tenure")

# Renaming a column
final_df = sorted_employees.rename(columns={"tenure": "years_in_company"})

print(final_df)

### END CODE HERE ###
````
