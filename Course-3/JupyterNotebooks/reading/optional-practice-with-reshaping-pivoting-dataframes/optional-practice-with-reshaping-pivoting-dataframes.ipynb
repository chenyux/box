## Exercise 1: Pivot the dataframe

You have a dataframe of temperatures in various cities. The columns of the dataframe are "date", "city" and "temperature". You want to pivot the dataframe so each column corresponds to a city and each row corresponds to a date and each entry in the cell is the temperature that corresponds to the date and city in the respective row and column.

### Instructions

- Run the cell below to print out the current dataframe to familiarize yourself with what you have.
- Pivot the dataframe so the rows correspond to dates and columns to cities.
- Print the final result (pivoted dataframe) to check for consistency.

````python
import pandas as pd

data = {
    "date": ["2023-01-01", "2023-01-01", "2023-01-01",
             "2023-01-02", "2023-01-02", "2023-01-02",
             "2023-01-03", "2023-01-03", "2023-01-03"],
    "city": ["New York", "Los Angeles", "Chicago",
             "New York", "Los Angeles", "Chicago",
             "New York", "Los Angeles", "Chicago"],
    "temperature": [5, 20, -3,
                    6, 22, -2,
                    4, 19, -5]
}

df = pd.DataFrame(data)

### START CODE HERE ###

print(df)

### END CODE HERE ###
````

**Expected output:**

```
city        Chicago  Los Angeles  New York
date                                      
2023-01-01       -3           20         5
2023-01-02       -2           22         6
2023-01-03       -5           19         4
```

## SOLUTION:

```python
pivoted_df = df.pivot(index="date", columns="city", values="temperature")
print(pivoted_df)
```
